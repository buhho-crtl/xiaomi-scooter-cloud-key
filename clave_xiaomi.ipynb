# Xiaomi key for your scooter

Gets, out of your Xiaomi account, the key the app asks for to connect to the scooter.

1. **First sign in to Xiaomi** in this browser: open <https://account.xiaomi.com> in another tab and sign in with the account you use in Mi Home.
2. Here, choose *Runtime → Run all*.
3. A Xiaomi link will show up below. Open it and accept, then come back here.
4. Copy the 64 characters next to `KEY:` and paste them into the app, with the PIN you set for the scooter in Mi Home.

If Xiaomi's page says "Invalid request" or "QR code has expired", you were not signed in yet: sign in at account.xiaomi.com and run again.

This runs on a Google machine tied to your Google account and is wiped when you close the tab. The program only talks to Xiaomi's servers.
