# Xiaomi key for your scooter

1. Choose *Runtime → Run all*.
2. A Xiaomi link will show up below. Open it, sign in **on Xiaomi's page** and come back here.
3. Copy the 64 characters next to `KEY:` and paste them into the app, with the PIN you set for the scooter in Mi Home.

This runs on a Google machine tied to your Google account and is wiped when you close the tab. The program only talks to Xiaomi's servers.
